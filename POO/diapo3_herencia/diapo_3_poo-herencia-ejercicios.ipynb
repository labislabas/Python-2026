#### **Ejercicio 1: Animales y Sonidos (`Mascota` -> `Perro`)**
* **Enunciado:** Crea una clase base `Mascota` que reciba el atributo `nombre` en su constructor y tenga un método `emitir_sonido()` con un mensaje genérico. Luego, crea la subclase `Perro` que herede de `Mascota` e implemente su propio método `emitir_sonido()` que retorne `'¡Guau guau!'`.
* **Tips:**
  1. Define `__init__(self, nombre)` en `Mascota` para guardar el nombre.
  2. En `Perro`, al no agregar atributos nuevos, heredará automáticamente el constructor de la clase base.

```python
# desarrollo del codigo
class Mascota:
    def __init__(self, nombre: str):
        # Guardamos el atributo 'nombre' en la instancia base
        self.nombre = nombre

    def emitir_sonido(self) -> str:
        # Método genérico que puede ser sobrescrito por las subclases
        return f"{self.nombre} hace un sonido genérico."


class Perro(Mascota):
    # Al no definir un nuevo __init__, hereda automáticamente
    # el constructor de Mascota.

    def emitir_sonido(self) -> str:
        # Sobrescribimos el método para devolver el ladrido específico
        return "¡Guau guau!"

class Gato(Mascota):
    # Al no definir un nuevo __init__, hereda automáticamente
    # el constructor de Mascota.

    def emitir_sonido(self) -> str:
        return "¡Miau miau!"

#

# 1. Instancia de la clase base Mascota
mascota_generica = Mascota("Firulais")
print(f"Nombre: {mascota_generica.nombre}")
print(f"Sonido: {mascota_generica.emitir_sonido()}")
# Salida:
# Nombre: Firulais
# Sonido: Firulais hace un sonido genérico.

print("-" * 30)

# 2. Instancia de la subclase Perro
mi_perro = Perro("Rex")
print(f"Nombre: {mi_perro.nombre}")          # Heredado automáticamente de Mascota
print(f"Sonido: {mi_perro.emitir_sonido()}")   # Método sobrescrito en Perro
# Salida:
# Nombre: Rex
# Sonido: ¡Guau guau!
```

#### **Ejercicio 2: Electrodomésticos y Consumo (`Electrodomestico` -> `Lavadora`)**
* **Enunciado:** Diseña una clase base `Electrodomestico` con atributos `marca` y `potencia_watts`. Crea la subclase `Lavadora` que agregue `carga_kg` y el método `calcular_consumo_diario(horas)` que devuelva el consumo en kWh usando la fórmula: \\(\frac{\text{potencia\_watts} \cdot \text{horas}}{1000}\\).
* **Tips:**
  1. Usa `super().__init__(marca, potencia_watts)` dentro del constructor de `Lavadora`.
  2. Asegúrate de dividir entre 1000 para convertir Watts a KiloWatts (kWh).

---

```python
# desarrollo del codigo

class Electrodomestico:
    def __init__(self, marca: str, potencia_watts: float):
        self.marca = marca
        self.potencia_watts = potencia_watts

class Lavadora(Electrodomestico):
    def __init__(self, marca: str, potencia_watts: float, carga_kg: float):
        super().__init__(marca, potencia_watts)
        self.carga_kg = carga_kg

    def Calcular_consumo_diario_en_horas(self, horas: float):
        return (self.potencia_watts * horas) / 1000

Lavadora_lia = Lavadora('Toshiba', 4500, 15)
print(f'La lavadora marca {Lavadora_lia.marca} de {Lavadora_lia.carga_kg}kg de capaciadad gasta {Lavadora_lia.Calcular_consumo_diario_en_horas(1)}kWh a la hora')
```

#### **Ejercicio 3: Permisos de Usuario (`Usuario` -> `UsuarioAdministrador`)**
* **Enunciado:** Crea una clase `Usuario` con `nombre_usuario` y `email`. Luego, crea la subclase `UsuarioAdministrador` con una lista de `permisos` (ej. `['crear', 'eliminar', 'editar']`) y un método `tiene_permiso(permiso)` que devuelva `True` o `False`.
* **Tips:**
  1. Utiliza el operador `in` de Python (`permiso in self.permisos`) para verificar la existencia del permiso.
  2. En el constructor del administrador, permite recibir la lista o asigna una por defecto.

```python
# desarrollo del codigo

class Usuario: 
    def __init__(self, nombre_usuario: str, email: str):
        self.nombre_usurio = nombre_usuario
        self.email = email

class UsarioAdministrador(Usuario):
    def __init__(self, nombre_usuario, email, permisos = ('crear', 'eliminar', 'editar')):
        super().__init__(nombre_usuario, email)
        self.permisos = permisos

    def Tiene_permiso(self, permisos):
        return permisos in self.permisos

administrador = UsarioAdministrador('Helber', 'socrandse@gmail.com')
print(f'Usuario = {administrador.nombre_usurio} | Email = {administrador.email}')
print(f'Usted tiene permiso para "crear" = {administrador.Tiene_permiso('crear')}')
print(f'Usted tiene permiso para "eliminar" = {administrador.Tiene_permiso('eliminar')}')
print(f'Usted tiene permiso para "clonar" = {administrador.Tiene_permiso('clonar')}')
print(f'Usted tiene permiso para "editar" = {administrador.Tiene_permiso('editar')}')
```

#### **Ejercicio 4: Registro Escolar (`PersonaEscolar` -> `Estudiante`)**
* **Enunciado:** Modela `PersonaEscolar` con `nombre` e `id_identificacion`. Crea la subclase `Estudiante` con una lista de `calificaciones` y los métodos `agregar_calificacion(nota)` y `calcular_promedio()`. Si no hay calificaciones, debe retornar `0.0`.
* **Tips:**
  1. Inicializa `self.calificaciones = []` en el constructor de `Estudiante`.
  2. Usa `sum(self.calificaciones)` y `len(self.calificaciones)`, verificando antes que la lista no esté vacía para evitar errores de división por cero (`ZeroDivisionError`).

---

```python
# desarrollo del codigo

class PersonaEscolar:
    def __init__(self, nombre: str, id_identificacion: int):
        self.nombre = nombre
        self.id_identificacion = id_identificacion

class Estudiante(PersonaEscolar):
    def __init__(self, nombre, id_identificacion):
        super().__init__(nombre, id_identificacion)
        self.calificaciones = []

    def agregar_calificacion(self, nota):
        self.calificaciones.append(nota)

    def calcular_promedio(self):
        if len(self.calificaciones) > 0: 
            return sum(self.calificaciones) / len(self.calificaciones)
        else: 
            return 0.0

estudiante_num_2 = Estudiante('Julien', 15)

estudiante_num_2.agregar_calificacion(1.9)
estudiante_num_2.agregar_calificacion(5.3)
estudiante_num_2.agregar_calificacion(6.9)

print(f'El promedio de {estudiante_num_2.nombre} fue: {estudiante_num_2.calcular_promedio()}')
```

#### **Ejercicio 5: Sensor de Temperatura (`Sensor` -> `SensorTemperatura`)**
* **Enunciado:** Diseña la clase `Sensor` con `ubicacion` y `lectura_actual`. Crea la subclase `SensorTemperatura` con el umbral `temperatura_maxima` y el método `evaluar_alerta()` que devuelva `'⚠️ ¡ALERTA: Temperatura crítica!'` si la lectura supera el límite, o `'✅ Temperatura dentro del rango normal.'` en caso contrario.
* **Tips:**
  1. Crea un método `actualizar_lectura(nuevo_valor)` en la clase base.
  2. Usa un condicional `if / else` dentro de `evaluar_alerta()` comparando `self.lectura_actual > self.temperatura_maxima`.

---

```python
# desarrollo del codigo

class Sensor:
    def __init__(self, ubicacion: str, lectura_actual: float):
        self.ubicacion = ubicacion
        self.lectura_actual = lectura_actual

    def actualizar_lectura(self, nuevo_valor: float):
        self.nuevo_valor = nuevo_valor

class SensorTemperatura(Sensor):
    def __init__(self, ubicacion, lectura_actual, temperatura_maxima: float):
        super().__init__(ubicacion, lectura_actual)
        self.temperatura_maxima = temperatura_maxima

    def evaluar_alerta(self):
        if self.lectura_actual > self.temperatura_maxima: 
            return '!ALERTA: Temperatura critica!'
        else:
            return 'Temperatura dentro del rango normal'

sensol_2 = SensorTemperatura('Planta baja', 45.5, 50.5)
sensol_3 = SensorTemperatura('Planta alta', 54.4, 50.5)

print(f'{sensol_2.evaluar_alerta()}, la temperatura del sensor de la {sensol_2.ubicacion} es de {sensol_2.lectura_actual}')
print(f'{sensol_3.evaluar_alerta()} La temperatura del sensor de la {sensol_3.ubicacion} es de {sensol_3.lectura_actual}')
```


#### **Ejercicio 6: Tienda Online (`Producto` -> `ProductoEnOferta`)**
* **Enunciado:** Crea la clase `Producto` con `nombre` y `precio_base`, y el método `obtener_precio_final()`. Crea la subclase `ProductoEnOferta` con `porcentaje_descuento` (ej. 20 para 20%) y sobrescribe `obtener_precio_final()` aplicando la rebaja.
* **Tips:**
  1. Aplica la fórmula: \\(\text{precio\_base} \cdot \left(1 - \frac{\text{porcentaje\_descuento}}{100}\right)\\).
  2. Conserva el mismo nombre del método (`obtener_precio_final`) para mantener el comportamiento polimórfico.




```python
# desarrollo del codigo

class Producto: 
    def __init__(self, nombre: str, precio_base: float):
        self.nombre = nombre
        self.precio_base = precio_base

    def obtener_precio_final(self):
        return self.precio_base

class ProductosEnOferta(Producto):
    def __init__(self, nombre, precio_base, porcentaje_descuento: float):
        super().__init__(nombre, precio_base)
        self.porcentaje_descuento = porcentaje_descuento

    def obtener_precio_final(self):
        return self.precio_base * (1 - self.porcentaje_descuento / 100)
    
compra_1 = ProductosEnOferta('mouse logitech g203', 25000, 5)
compra_2 = ProductosEnOferta('teclado aula f75', 82000, 20)
print(f'El {compra_1.nombre} tras aplicarle el descuento del {compra_1.porcentaje_descuento}% queda en: {compra_1.obtener_precio_final()}')
print(f'El {compra_2.nombre} tras aplicarle el descuento del {compra_2.porcentaje_descuento}% queda en: {compra_2.obtener_precio_final()}')
```

#### **Ejercicio 7: Geometría 3D (`Circulo` -> `Cilindro`)**
* **Enunciado:** Desarrolla la clase `Circulo` con el atributo `radio` y el método `area()` (\\(\pi \cdot r^2\\)). Crea la subclase `Cilindro` que agregue `altura` e implemente el método `volumen()` reutilizando el método `area()` del padre (\\(\text{volumen} = \text{area\_base} \cdot \text{altura}\\)).
* **Tips:**
  1. Importa `math` y utiliza la constante `math.pi`.
  2. Dentro de `volumen()`, invoca `self.area()` o `super().area()` para obtener la superficie de la base circular sin repetir la fórmula.

```python
# desarrollo del codigo
import math

class Circulo:
    def __init__(self, radio: float):
        self.radio = radio

    def area(self: float):
        return math.pi * (self.radio ** 2)

class Cilindro(Circulo):
    def __init__(self, radio, altura: float):
        super().__init__(radio)
        self.altura = altura

    def volumen(self: float):
        return self.area() * self.altura

operacion_1 = Cilindro(34.4, 37.5)
print(f'El volumen de su cilindro es de: {operacion_1.volumen()}')
```
