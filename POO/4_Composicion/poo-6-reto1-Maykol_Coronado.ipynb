## **Reto 1: Jugador y Arma (Videojuego básico)**

**Detalle:**

Diseña un sistema para un juego por turnos:

1. Crea una clase `Arma` con los atributos `nombre` y `danio`. Debe tener un método `atacar()` que devuelva el string: `f"{self.nombre} causa {self.danio} puntos de daño."`.
2. Crea una clase `Guerrero` con los atributos `nombre` y `puntos_vida`. Un guerrero se crea sin arma (`self.arma = None`).
3. Agrega al `Guerrero` los métodos:
* `equipar_arma(arma)`: asigna un objeto `Arma` al guerrero.
* `realizar_ataque()`: si tiene un arma equipada, imprime `f"{self.nombre} ataca: "` seguido del resultado de `arma.atacar()`. Si no tiene arma, imprime `f"{self.nombre} ataca con los puños (1 punto de daño)."`.



```python
class Arma():
    """ Molde con el cual se asignan el nombre y dano del 'arma' """
    def __init__(self, nombre: str, danio:float):
        self.nombre = nombre
        self.danio = danio
    """ Describe el ataque realizada con el arma. """
    def atacar(self):
        return f'{self.nombre} causa {self.danio} puntos de dano.'

class Guerrero():
    """ Personaje al cual se le asignaran atributos nombre, puntos de vida, ademas de posteriormente ser una isntancia de 'usuario' """
    def __init__(self, nombre: str, puntos_vida: float):
        self.nombre = nombre
        self.puntos_vida = puntos_vida
        self.arma = None

    """ Metodo con el cual posteriormente se podra agregar armas al "Guerrero" """
    def equipar_arma(self, arma: Arma):
        self.arma = arma

    """ Metodo con el cual se verifica que el Guerrero tenga un arma, de caso contrario se le agrega un string predeterminado"""
    def realizar_ataque(self):
        if self.arma is not None:
            print(f'{self.nombre} ataca: {self.arma.atacar()}')
        else:
            print(f'{self.nombre} ataca con los punos (1 punto de dano.)')

usuario = Guerrero('Pessel', 8)

print(f'El guerrero {usuario.nombre} cuenta con {usuario.puntos_vida} puntos de vida.')

usuario.equipar_arma(Arma('Machete', 6))
usuario.realizar_ataque()
```

Como se pedia en las instrucciones, el objetivo era crear un clase llamada Guerrero el cual pudiera usar metodos de la clase Arma y asi poder crear un personaje el cual se pudiera modificar el arma y el dano que infringe con esta misma, amdeas de contar con un nombre y puntos de vida.

## **Reto 2: Biblioteca y Libros (Colección de objetos)**

**Detalle:**

Modela el catálogo de una biblioteca pequeña:

1. Crea una clase `Libro` con los atributos `titulo` y `autor`. Agrega un método `descripcion()` que retorne: `f"'{self.titulo}' por {self.autor}"`.
2. Crea una clase `Biblioteca` con un nombre y una lista vacía para almacenar libros.
3. Agrega a `Biblioteca` los métodos:
* `agregar_libro(libro)`: añade una instancia de `Libro` a la lista interna.
* `mostrar_catalogo()`: recorre la lista e imprime cada libro usando el método `descripcion()` de cada uno. Si la lista está vacía, debe avisar que no hay libros registrados.

```python
class Libro:
    """ Molde con el cual se asignan el titulo y autor """
    def __init__(self, titulo: str, autor: str):
        self.titulo = titulo
        self.autor = autor

    def descripcion(self):
        return f"'{self.titulo}' por {self.autor}"


class Biblioteca:
    """ Clase la cual se le asignara el atributo nombre, para posteriormente ser una isntancia de 'biblioteca'"""
    def __init__(self, nombre: str):
        self.nombre = nombre
        self.almacenar_libros = []

    """ Metodo con el cual posteriormente se podra agregar libros con su respectivo titulo y autor """
    def agregar_libro(self, libro: Libro):
        self.almacenar_libros.append(libro)

    """ Metodo en el que se verifica que existan libros en "almacenar_libros" """
    def mostrar_catalogo(self):
        if len(self.almacenar_libros) > 0:
            for libro in self.almacenar_libros:
                print(libro.descripcion())
        else:
            print('No hay libros registrados.')


biblioteca = Biblioteca('Biblioteca central de la Ulagos')
print(f'Los libros disponibles de {biblioteca.nombre} son: ')
biblioteca.agregar_libro(Libro('La divina proyectada', 'Dun kafku'))
biblioteca.agregar_libro(Libro('El lamento cordillerano', 'Hall Lirksen'))
biblioteca.mostrar_catalogo()
```

Como se pedia en las instrucciones, el objetivo era crear un clase llamada Biblioteca el cual pudiera usar metodos de la clase Libro y asi poder crear un un biblioteca la cual se pudiera modificar los libros que hay en esta misma con su titulo y autor.

## **Reto 3: Factura e Ítems con Cantidad (Cálculo acumulativo)**

**Detalle:**

Modela la emisión de un comprobante de venta simple:

1. Crea una clase `ItemFactura` con los atributos `descripcion`, `precio_unitario` y `cantidad`. Debe tener un método `calcular_subtotal()` que retorne el total del ítem (`precio_unitario * cantidad`).
2. Crea una clase `Factura` con el atributo `numero_factura` y una lista de ítems.
3. Agrega a `Factura` los métodos:
* `agregar_item(descripcion, precio_unitario, cantidad)`: crea la instancia de `ItemFactura` dentro del método y la añade a la lista interna.
* `calcular_total()`: suma los subtotales de todos los ítems de la lista.
* `imprimir_resumen()`: muestra el número de factura, el desglose de cada ítem (descripción, cantidad y subtotal) y el total final.



```python
class ItemFactura:
    def __init__(self, descripcion: str, precio_unitario: float, cantidad: int):
        self.descripcion = descripcion
        self.precio_unitario = precio_unitario
        self.cantidad = cantidad

    def calcular_subtotal(self):
        return self.precio_unitario * self.cantidad


class Factura:
    def __init__(self, numero_factura: int):
        self.numero_factura = numero_factura
        self.lista_items = []

    def agregar_item(self, descripcion: str, precio_unitario: float, cantidad: int):
        compra = ItemFactura(descripcion, precio_unitario, cantidad)
        self.lista_items.append(compra)

    def calcular_total(self):
        total = 0
        for i in self.lista_items:
            total += i.calcular_subtotal()
        return total

    def imprimir_resumen(self):
        print(f'Número de factura: {self.numero_factura}')
        for i in self.lista_items:
            print(f'Descripción: {i.descripcion} | Cantidad: {i.cantidad} | Subtotal: {i.calcular_subtotal()}')
        print(f'Total: {self.calcular_total()}')


factura = Factura(76)
factura.agregar_item('Lapiz', 1500, 2)
factura.agregar_item('Cuaderno', 2900, 1)
factura.imprimir_resumen()
```
