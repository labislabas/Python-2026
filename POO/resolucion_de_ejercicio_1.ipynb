```python
# 1. Definicion de la CLASE (El molde)
class Dog:
    """Un intento sencillo de modelar un perro."""

    def __init__(self, name, age, race, weight):
        """Inicializa los atributos de nombre y edad."""
        self.name = name    # Atributo de instancia para el nombre
        self.age = age  #Atributo de instancia para la edad
        self.race = race    #Atributo de instancia para la raza
        self.weight = weight    #Atributo de instancia para el peso

    def sit(self):
        """Simula un perro sentándose en respuesta a una orden."""
        print(f"{self.name} is now sitting.")

    def rueda(self):
        """Simula hacer la croqueta en respuesta a una orden."""
        print(f"{self.name} rolled over!")

    def saltar(self):
        """Simula un perro saltando."""
        print(f"{self.name} salta muy alto.")

    def ladrar(self):
        """Simula un perro ladrando."""
        print(f"{self.name} ladra muy fuerte.")

# --- Demostración de uso del programa ---

# 1. Creamos una instancia específica de la clase Dog
my_dog = Dog("willie", 6, 'Pastor alemán', 28.2)

# 2. Accedemos e imprimimos sus atributos
print(f"Mi perro se llama {my_dog.name}.")
print(f"Mi perro tiene {my_dog.age} años.")
print(f"Mi perro es de una raza llamda {my_dog.race}")
print(f"Mi perro pesa {my_dog.weight}")

# 3. Llamamos a los métodos del objeto 
my_dog.sit()
my_dog.rueda()
my_dog.saltar()
my_dog.ladrar()

# 4. Creación de una segunda instancia independiente
your_dog = Dog("Lucy", 3, 'Chihuahua', 7.1)
print(f"\nMi otro perro se llama {your_dog.name}.")
print(f"Mi otro perro tiene {your_dog.age} años.")
print(f"Mi otro perro es de raza {your_dog.race}")
print(f"Mi otro perro pesa {your_dog.weight}")
your_dog.sit()
your_dog.rueda()
your_dog.saltar()
your_dog.ladrar()

```

```python
# 1. Definicion de la CLASE (El molde)
class Loro:
    """Un intento sencillo de modelar un loro."""

    def __init__(self, name, age, color, height):
        """Inicializa los atributos de nombre y edad."""
        self.name = name    # Atributo de instancia para el nombre
        self.age = age  #Atributo de instancia para la edad
        self.color = color    #Atributo de instancia para el color
        self.height = height    #Atributo de instancia para la estatura (en cm)

    def volar(self):
        """Simula un loro volando en respuesta a una orden."""
        print(f"{self.name} esta volando muy alto pese a su estatura.")

    def imitar(self):
        """Simula un loro imitando a un humano en respuesta a una orden."""
        print(f"{self.name} te esta imitando.")

    def acicalarse(self):
        """Simula un loro acicalandose."""
        print(f"{self.name} esta asicalandose sus plumas.")

    def trepar(self):
        """Simula un loro escalando un arbol con sus patas"""
        print(f"Mira, {self.name} esta trepando ese arbol solo con sus pies.")

# --- Demostración de uso del programa ---

# 1. Creamos una instancia específica de la clase loro
my_loro = Loro("Kassel", 23, 'verde claro', '34 centimetros')

# 2. Accedemos e imprimimos sus atributos
print(f"Mi loro se llama {my_loro.name}.")
print(f"Mi loro tiene {my_loro.age} años.")
print(f"Mi loro es de color {my_loro.color}")
print(f"Mi loro mide {my_loro.height}")

# 3. Llamamos a los métodos del objeto 
my_loro.volar()
my_loro.imitar()
my_loro.acicalarse()
my_loro.trepar()

# 4. Creación de una segunda instancia independiente
your_loro = Loro("Hessel", 18, 'verde oscuro', '26 centimetros')
print(f"\n Mi loro se llama {your_loro.name}.")
print(f"Mi loro tiene {your_loro.age} años.")
print(f"Mi loro es de color {your_loro.color}")
print(f"Mi loro mide {your_loro.height}")
your_loro.volar()
your_loro.imitar()
your_loro.acicalarse()
your_loro.trepar()
```

```python
# 1. Definicion de la CLASE (El molde)
class Cangrejo:
    """Un intento sencillo de modelar un cangrejo."""

    def __init__(self, name, age, color, species):
        """Inicializa los atributos de nombre y edad."""
        self.name = name    # Atributo de instancia para el nombre
        self.age = age  #Atributo de instancia para la edad
        self.color = color    #Atributo de instancia para el color
        self.species = species    #Atributo de instancia para su especie (en cm)

    def depredador(self):
        """Simula un Cangrejo cazando."""
        print(f"{self.name} al parecer tiene hambre, y esta intentando cazar a caracoles.")

    def territorial(self):
        """Simula un Cangrejo siendo agresivo."""
        print(f"{self.name} esta siendo agresivo con los demas peces, ya que han entrado en su territorio.")

    def no_tan_acuatico(self):
        """Simula un Cangrejo saliendo a la superficie, ya que requiere aire."""
        print(f"{self.name} ha pasado mucho tiempo en el agua, por lo que tiene que salir a tomar aire.")

    def pinzas(self):
        """Simula un Cangrejo escalando una planta con sus pinzas"""
        print(f"Mira, {self.name} esta trepando esa planta con sus pinzas.")

# --- Demostración de uso del programa ---

# 1. Creamos una instancia específica de la clase Cangrejo
my_Cangrejo = Cangrejo("Coclea", 2, 'negro con sus pinzas negras', 'cangrejo de pinzas rojas')

# 2. Accedemos e imprimimos sus atributos
print(f"Mi Cangrejo se llama {my_Cangrejo.name}.")
print(f"Mi Cangrejo tiene {my_Cangrejo.age} años.")
print(f"Mi Cangrejo es de color {my_Cangrejo.color}")
print(f"Mi Cangrejo es de una especie llamada: {my_Cangrejo.species}")

# 3. Llamamos a los métodos del objeto 
my_Cangrejo.depredador()
my_Cangrejo.territorial()
my_Cangrejo.no_tan_acuatico()
my_Cangrejo.pinzas()

# 4. Creación de una segunda instancia independiente
your_Cangrejo = Cangrejo("Davila", 1, 'naranjo oscuro', 'cangrejo de pinnzas rojas')
print(f"\n Mi Cangrejo se llama {your_Cangrejo.name}.")
print(f"Mi Cangrejo tiene {your_Cangrejo.age} años.")
print(f"Mi Cangrejo es de color {your_Cangrejo.color}")
print(f"Mi Cangrejo mide {your_Cangrejo.species}")
your_Cangrejo.depredador()
your_Cangrejo.territorial()
your_Cangrejo.no_tan_acuatico()
your_Cangrejo.pinzas()
```
